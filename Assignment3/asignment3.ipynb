```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)

employees.to_csv("employees.csv", index=False)

employees
```

```python
# a) Shape of the DataFrame

employees.shape

```

```python

employees.info()
```

```python

employees.describe()

```

```python

print("First 5 rows:")
display(employees.head(5))

print("Last 3 rows:")
display(employees.tail(3))
```

```python
# e) Statistics

print("i. Average salary =", employees["Salary"].mean())
print("ii. Total bonus =", employees["Bonus"].sum())
print("iii. Youngest employee's age =", employees["Age"].min())
print("iv. Highest performance rating =", employees["Rating"].max())
```

```python

employees_sorted = employees.sort_values("Salary", ascending=False)

employees_sorted
```

```python

def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(performance)

employees
```

```python

employees.isnull().sum()
```

```python
# i) Rename Employee_ID to ID

employees.rename(columns={"Employee_ID": "ID"}, inplace=True)

employees
```

```python
# j) Find employees with more than 5 years of experience

employees[employees["Years_of_Experience"] > 5]
```

```python
# j) Find employees who belong to the IT department

employees[employees["Department"] == "IT"]
```

```python
# k) Add Tax column

employees["Tax"] = employees["Salary"] * 0.10

employees
```

```python

employees.to_csv("employees_modified.csv", index=False)

print("Modified DataFrame saved as employees_modified.csv")
```
