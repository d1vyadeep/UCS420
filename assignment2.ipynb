```python
roll_no = input("Enter your roll number: ")

digits = []
for digit in roll_no:
    digits.append(int(digit))

print("Roll number:", roll_no)
print("Digits:", digits)
```

```python
L = []

for digit in digits:
    L.append(digit * 10)

print("L =", L)
```

```python
L.append(50)
print("After append(50):", L)

L.insert(2, 70)
print("After insert(2, 70):", L)
```

```python
L.remove(50)
print("After remove(50):", L)

L.pop(2)
print("After pop(2):", L)
```

```python
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average =", greater_than_average)
```

```python
import random

random.seed(int(roll_no))

numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

print(numbers)

```

```python
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
odd_numbers = []

for number in numbers:
    if number % 2 != 0:
        odd_numbers.append(number)

print("Odd numbers =", odd_numbers)
print("Count of odd numbers =", len(odd_numbers))
```

```python
even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print("Even numbers =", even_numbers)
print("Count of even numbers =", len(even_numbers))
```

```python
prime_numbers = []

for number in numbers:
    if number > 1:
        prime = True

        for i in range(2, number):
            if number % i == 0:
                prime = False
                break

        if prime:
            prime_numbers.append(number)

print("Prime numbers =", prime_numbers)
print("Count of prime numbers =", len(prime_numbers))
```

```python
prime_numbers_comprehension = [
    number
    for number in numbers
    if number > 1 and all(number % i != 0 for i in range(2, int(number ** 0.5) + 1))
]

print("Prime numbers using list comprehension =", prime_numbers_comprehension)
```

```python
most_number = numbers[0]
most_count = numbers.count(numbers[0])

for number in numbers:
    count = numbers.count(number)

    if count > most_count:
        most_number = number
        most_count = count

print("Most frequent number =", most_number)
print("Number of times it occurs =", most_count)
```

```python
first_8_digits = digits[:8]

A = set()
B = set()

for digit in first_8_digits:
    A.add(digit * 7)
    B.add(digit * 9)

print("A =", A)
print("B =", B)
```

```python
print("Union =", A.union(B))
```

```python
print("Intersection =", A.intersection(B))

```

```python
print("A - B =", A.difference(B))
print("B - A =", B.difference(A))
```

```python
print("Symmetric difference =", A.symmetric_difference(B))
```

```python
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```python
X = int(input("Enter X to remove from A: "))

A.discard(X)

print("A after discard =", A)
```
